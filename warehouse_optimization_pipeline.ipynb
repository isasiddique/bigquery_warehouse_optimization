```python
import os
import json
import random
import numpy as np
import pandas as pd

# 1. INITIALIZE MASTER CLOUD WAREHOUSE STORAGE DIRECTORIES
os.makedirs('warehouse_logs', exist_ok=True)

# Define target architectural schema states (Simulating flat legacy tables vs optimized partitioned arrays)
warehouse_table_profiles = [
    {"table_type": "LEGACY_UNPARTITIONED_FULL_SCAN", "ingest_year": 2024, "scanned_bytes_gb": 450.5, "query_duration_sec": 42},
    {"table_type": "LEGACY_UNPARTITIONED_FULL_SCAN", "ingest_year": 2025, "scanned_bytes_gb": 820.2, "query_duration_sec": 89}, # Billing runaway hazard
    {"table_type": "OPTIMIZED_PARTITIONED_CLUSTERED", "ingest_year": 2026, "scanned_bytes_gb": 12.4, "query_duration_sec": 2},
    {"table_type": "OPTIMIZED_PARTITIONED_CLUSTERED", "ingest_year": 2026, "scanned_bytes_gb": 8.1, "query_duration_sec": 1}
]

# 2. RUN BIGQUERY STORAGE COMPLIANCE AUDITOR (5,000 continuous schema sweeps)
np.random.seed(42)
n_queries = 5000
warehouse_records = []
optimized_partitions = 0
billing_runaways = 0

print("--- EXECUTING COLUMNS WAREHOUSE RETRIEVAL SWEEP ---")
for i in range(1, n_queries + 1):
    query_id = f"BQ-{i:05d}"

    # Ingest a live query execution snapshot from an enterprise logging cluster
    query_snapshot = random.choice(warehouse_table_profiles)
    t_type = query_snapshot["table_type"]
    year = query_snapshot["ingest_year"]
    bytes_scanned = query_snapshot["scanned_bytes_gb"]
    duration = query_snapshot["query_duration_sec"]

    # Cloud FinOps Governance Check: Isolate high-cost legacy unpartitioned scans
    # Enterprise SLA targets query footprint limits < 100GB per single execution plan
    is_scan_economical = bytes_scanned < 100.0
    is_execution_efficient = duration <= 10

    if is_scan_economical and is_execution_efficient:
        governance_state = "FINOPS_COMPLIANT_SCHEMA"
        zeta_action = "RETAIN_STORAGE_CONFIGURATION"
        http_code = 200
        optimized_partitions += 1
    else:
        governance_state = "CRITICAL_BILLING_RUNAWAY_HAZARD"
        zeta_action = "EXECUTE_AUTOMATED_BIGQUERY_PARTITION_CLUSTER_MIGRATION"
        http_code = 402  # Payment/Cost Exception Required
        billing_runaways += 1

    warehouse_records.append(
        f"{query_id} | {t_type} | {year} | {bytes_scanned:.1f} | {duration} | "
        f"{governance_state} | {zeta_action} | {http_code}\n"
    )

# Save the immutable operational bigquery warehouse audit log
with open('warehouse_logs/bigquery_warehouse_audit.log', 'w') as f:
    f.writelines(warehouse_records)

# 3. TRANSFORMATIONAL STRUCTURE: REBUILD FLAT SECTORS INTO NESTED ALIGNMENT SCHEMAS
df_bq = pd.read_csv('warehouse_logs/bigquery_warehouse_audit.log', sep='|',
                    names=['QueryID', 'TableType', 'Year', 'BytesScanned', 'Duration', 'State', 'Action', 'Status'])

for col in df_bq.columns:
    df_bq[col] = df_bq[col].astype(str).str.strip()

nested_zeta_json = []
for idx, row in df_bq.head(2).iterrows():
    json_record = {
        "bigQueryTransactionIdentifier": row['QueryID'],
        "storageAssetMetadata": {
            "evaluatedTableSchemaType": row['TableType'],
            "ingestYearPartition": int(row['Year'])
        },
        "queryPlanExecutionTelemetry": {
            "dataFootprintScannedGB": float(row['BytesScanned']),
            "queryRuntimeDurationSeconds": int(row['Duration'])
        },
        "cloudFinOpsGovernance": {
            "billingComplianceStatus": row['State'],
            "ZetaAgentActionDirective": row['Action'],
            "infrastructureResponseStatusCode": int(row['Status'])
        }
    }
    nested_zeta_json.append(json_record)

with open('warehouse_logs/zeta_bigquery_manifest.json', 'w') as json_file:
    json.dump(nested_zeta_json, json_file, indent=4)

print("\n--- ZETA BIGQUERY FINOPS MONITOR ACTIVE ---")
print(f"Total Petabyte-Scale Warehouse Query Plans Audited: {n_queries}")
print(f"Query Runs Matching FinOps-Compliant Optimized Baselines: {optimized_partitions}")
print(f"Query Runs Intercepted with Volatile Data Scanning Spikes: {billing_runaways}\n")
print("--- AUTONOMOUS ZETA AGENT BIGQUERY SCHEMA ---")
print(json.dumps(nested_zeta_json, indent=2))
```
